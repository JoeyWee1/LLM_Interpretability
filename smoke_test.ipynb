```python
#!/usr/bin/env python3
import random

WIN_LINES = [
    (0, 1, 2), (3, 4, 5), (6, 7, 8),  # rows
    (0, 3, 6), (1, 4, 7), (2, 5, 8),  # cols
    (0, 4, 8), (2, 4, 6)              # diagonals
]

def render(board):
    def cell(i):
        return board[i] if board[i] != " " else str(i + 1)
    rows = []
    for r in range(0, 9, 3):
        rows.append(f" {cell(r)} | {cell(r+1)} | {cell(r+2)} ")
    return "\n---+---+---\n".join(rows)

def winner(board):
    for a, b, c in WIN_LINES:
        if board[a] != " " and board[a] == board[b] == board[c]:
            return board[a]
    return None

def is_draw(board):
    return all(x != " " for x in board) and winner(board) is None

def available_moves(board):
    return [i for i, v in enumerate(board) if v == " "]

def ask_move(board, player):
    while True:
        s = input(f"Player {player}, pick a square (1-9): ").strip()
        if not s.isdigit():
            print("Please enter a number 1-9.")
            continue
        idx = int(s) - 1
        if idx < 0 or idx > 8:
            print("Number must be 1-9.")
            continue
        if board[idx] != " ":
            print("That square is taken.")
            continue
        return idx

def cpu_move(board):
    # Random CPU; swap in smarter logic later if you want (minimax etc.)
    return random.choice(available_moves(board))

def main():
    print("Tic-Tac-Toe (CLI)")
    print("Choose mode:")
    print("  1) Two players")
    print("  2) Vs CPU (random)")
    mode = ""
    while mode not in {"1", "2"}:
        mode = input("Enter 1 or 2: ").strip()

    board = [" "] * 9
    player = "X"

    while True:
        print("\n" + render(board) + "\n")

        if mode == "2" and player == "O":
            move = cpu_move(board)
            print(f"CPU (O) plays: {move + 1}")
        else:
            move = ask_move(board, player)

        board[move] = player

        w = winner(board)
        if w:
            print("\n" + render(board))
            print(f"\n{w} wins!")
            break
        if is_draw(board):
            print("\n" + render(board))
            print("\nIt's a draw!")
            break

        player = "O" if player == "X" else "X"

if __name__ == "__main__":
    main()
9
```
